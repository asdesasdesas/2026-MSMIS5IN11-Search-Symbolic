# F2 – Argumentation à valeurs pour la décision médicale partagée

## Cas clinique : arthrose sévère du genou

L'arthrose du genou est une maladie fréquente pouvant entraîner des douleurs chroniques, une diminution de la mobilité et une altération de la qualité de vie.

Dans certains cas, plusieurs options thérapeutiques sont possibles :

- chirurgie (prothèse du genou) ;
- traitement conservateur (kinésithérapie, activité physique adaptée, perte de poids, traitements antalgiques).

Le choix dépend non seulement de critères médicaux mais également des valeurs et des préférences du patient.

L'objectif du projet est de montrer comment les préférences d'un patient influencent la décision grâce à un Value-based Argumentation Framework (VAF).

---

# Théorie

## Cadre d'argumentation de Dung

Un cadre d'argumentation de Dung est composé :

- d'un ensemble d'arguments ;
- d'un ensemble d'attaques entre arguments.

Le modèle permet de déterminer quels arguments peuvent être acceptés simultanément.

Les sémantiques utilisées dans le projet sont :

- Grounded
- Preferred
- Stable
- Semi-stable
- Ideal

Le projet utilise principalement l'extension grounded pour l'interprétation des résultats.

## Value-based Argumentation Framework (VAF)

Le VAF étend le modèle de Dung en associant une valeur à chaque argument.

Trois valeurs sont étudiées :

- mobilité ;
- sécurité ;
- qualité de vie.

Chaque patient possède un ordre de préférence différent entre ces valeurs.

Une attaque n'est conservée que si la valeur portée par l'argument attaquant est au moins aussi importante que celle portée par l'argument attaqué.

Ainsi, deux patients recevant les mêmes informations médicales peuvent aboutir à des conclusions différentes.

---

# Valeurs étudiées

## Mobilité

Retrouver la capacité à marcher, monter des escaliers et réaliser les activités quotidiennes.

## Sécurité

Éviter les risques et complications liés à la chirurgie.

## Qualité de vie

Réduire durablement la douleur et améliorer le confort quotidien.

---

# Arguments du débat

| ID | Argument | Valeur |
|----|----------|---------|
| A1 | La chirurgie améliore significativement la mobilité à long terme. | Mobilité |
| A2 | La chirurgie comporte des risques de complications. | Sécurité |
| A3 | Le traitement conservateur permet d'éviter les risques liés à la chirurgie. | Sécurité |
| A4 | Le traitement conservateur peut être insuffisant pour retrouver une bonne mobilité. | Mobilité |
| A5 | La chirurgie permet souvent une réduction plus importante et plus durable de la douleur. | Qualité de vie |
| A6 | Le traitement conservateur permet souvent de contrôler la douleur sans chirurgie. | Qualité de vie |

---

# Attaques entre arguments

| Attaque | Justification |
|----------|----------|
| A1 → A3 | Retrouver sa mobilité soutient la chirurgie. |
| A3 → A1 | Éviter l'opération soutient le traitement conservateur. |
| A2 → A1 | Les risques chirurgicaux sont un argument contre l'opération. |
| A1 → A2 | Le gain de mobilité peut justifier certains risques. |
| A4 → A3 | Le traitement conservateur n'est pas toujours suffisant. |
| A5 → A6 | La chirurgie peut mieux réduire la douleur. |
| A6 → A5 | Le traitement conservateur peut déjà contrôler la douleur. |
| A5 → A2 | Une forte réduction de la douleur peut justifier certains risques chirurgicaux. |

---

# Profils de patients

## Audience 1

Mobilité > Qualité de vie > Sécurité

Patient qui souhaite avant tout retrouver sa mobilité et son autonomie physique.

## Audience 2

Sécurité > Qualité de vie > Mobilité

Patient qui souhaite avant tout éviter les risques liés à une opération.

## Audience 3

Qualité de vie > Mobilité > Sécurité

Patient qui souhaite principalement réduire la douleur et améliorer son confort quotidien.

# Résultats

## Référence Dung

Grounded = {A4}

Preferred :

- {A1, A4, A5}
- {A1, A4, A6}
- {A2, A4, A6}

Observation :

Sans préférences de valeurs, plusieurs ensembles d'arguments admissibles sont possibles.

## Priorité à la mobilité

Grounded = {A1, A4}

Conclusion :

La chirurgie est favorisée.

## Priorité à la sécurité

Grounded = {A2, A3, A4}

Conclusion :

Le traitement conservateur est favorisé.

## Priorité à la qualité de vie

Grounded = {A1, A4}

Conclusion :

La chirurgie est favorisée.

---

# Analyse

Le VAF modifie les résultats selon les priorités du patient.

Lorsque la sécurité devient la valeur principale, les arguments liés aux risques chirurgicaux deviennent dominants et le traitement conservateur est privilégié.

Les audiences « mobilité » et « qualité de vie » produisent les mêmes conclusions dans ce modèle.

Cela ne constitue pas une erreur : les attaques conservées sont identiques pour ces deux audiences, ce qui conduit aux mêmes extensions.

# Vérification indépendante

Le projet contient un second moteur nommé `verifier.py`.

Ce module recalcule indépendamment l'extension grounded.

Les tests vérifient automatiquement que :

- le moteur principal ;
- le vérificateur indépendant ;

produisent les mêmes résultats.

Résultat :

- 10 tests exécutés ;
- 10 tests validés.

Commande utilisée :

```bash
python -m unittest discover -s tests -v
```

# Performance

Résultats obtenus avec `benchmark.py`.

| Nombre d'arguments | Temps |
|-------------------|--------|
| 4 | 0.208 ms |
| 8 | 1.699 ms |
| 12 | 18.916 ms |
| 16 | 267.576 ms |

Observation :

Le temps d'exécution augmente rapidement avec la taille du graphe.

Cette évolution est cohérente avec l'énumération des sous-ensembles utilisée dans le calcul des sémantiques.

Le prototype est donc adapté à des graphes pédagogiques de petite taille mais pas à des instances de grande taille.

---

# Limites

- Cas clinique simplifié.
- Attaques modélisées à des fins pédagogiques.
- Graphes de petite taille uniquement.
- Pas d'intégration de Tweety.
- Profils de préférences illustratifs.
- Le système ne fournit pas de recommandation médicale réelle.
